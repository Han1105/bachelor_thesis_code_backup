```python
import numpy as np
import matplotlib.pyplot as plt

# 参数
L = 0.25  # 腔长，单位：m
z = np.linspace(0, L, 500)  # 腔内 z 轴坐标

# 高斯光束参数
w0 = 0.3e-3  # 光腰半径，单位：m（可以根据实际调整）
z_R = np.pi * w0**2 / (1065e-9)  # 瑞利长度
w_z = w0 * np.sqrt(1 + (z / z_R)**2)  # 各位置的光束半径

# 画图
plt.figure(figsize=(8, 3))
plt.plot(z, w_z * 1e3, 'r', label='Beam upper edge')  # 上边界（单位 mm）
plt.plot(z, -w_z * 1e3, 'r')                         # 下边界
plt.axvline(0, color='black')                        # 左镜 S1
plt.axvline(L, color='black')                        # 右镜 S2
plt.text(0, 0.5, 'S1 (Flat, waist)', ha='left')
plt.text(L, 0.5, 'S2 (Curved)', ha='right')
plt.title('Gaussian Beam in Flat-Curved Resonator')
plt.xlabel('z (m)')
plt.ylabel('Beam radius (mm)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
